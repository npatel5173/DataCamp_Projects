```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Print the dataset to see what it looks like
pd.set_option('display.max_columns', None)
print(netflix_df.loc[0:9, ['show_id', 'type', 'title', 'release_year', 'duration', 'genre']])
```

```python
# Filter the data for movies released from 1990s
filtered_df = netflix_df[
                (netflix_df['type'] == 'Movie') & 
                (netflix_df['release_year'] >= 1990) & 
                (netflix_df['release_year'] < 2000)
]
print(filtered_df)
# Find most frequent movie duration
plt.hist(filtered_df['duration'])
# Save answer as an integer called duration
duration = 100
```

```python
# Count the number of short action movies released from the 1990s
filtered_df2 = netflix_df[
                (netflix_df['genre'] == 'Action') & 
                (netflix_df['release_year'] >= 1990) & 
                (netflix_df['release_year']<=1999)
]
print(filtered_df2)

short_movie_count = 0
for label,row in filtered_df2.iterrows():
    if row['duration'] < 90:
        short_movie_count = short_movie_count + 1

# Save answer as an integer called short_movie_count
print(short_movie_count)
```
